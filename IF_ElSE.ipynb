### `Problem 1`: Write a program that will give you in hand monthly salary after deduction on CTC - HRA(10%), DA(5%), PF(3%) and taxes deduction as below:

> Salary(Lakhs) : Tax(%)

*   Below 5 : 0%
*   5-10 : 10%
*   10-20 : 20%
*   aboove 20 : 30%

````python
# Write code here
salary=int(input("Enter your salary in lakh"))

HRA=0.1
DA=0.05
PF=0.03

#Deduct HRA
salary=salary-salary*HRA
#Deduct DA
salary=salary-salary*DA
#Deduct PF
salary=salary-salary*PF


if salary<5:
  salary=salary
elif salary<10:
  salary=salary-salary*0.1
elif salary<20:
  salary=salary-salary*0.2
else:
  salary=salary-salary*0.3

salary=salary/12

print(f"Your monthly in hand monthly salary is {round(salary*100,2)}K")
````

### `Problem 6` - Find the factorial of a given number.

Write a program to use the loop to find the factorial of a given number.

The factorial (symbol: `!`) means to multiply all whole numbers from the chosen number down to 1.

For example: calculate the factorial of 5

```bash
5! = 5 × 4 × 3 × 2 × 1 = 120
```

Output:

```bash
120
```

````python
# Write code here

n=int(input("Enter a number"))
result=1
for i in range(n,0,-1):
  result*=i
print(result)
````

### `Problem 9`: Write a program that keeps on accepting a number from the user until the user enters Zero. Display the sum and average of all the numbers.

````python
# Write code here
i=1
nums=[]
sum=0
while i!=0:
  i=int(input("Enter a number"))
  nums.append(i)
  sum+=i

print(f"Sum of nums : {sum}")

#for avg
#count values
count=0
for i in nums:
  count+=1

print(f"Avg of nums : {sum/(count-1)}")
````
